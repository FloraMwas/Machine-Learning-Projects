```python
import tensorflow as tf
from tensorflow.keras import layers,models
```

```python
DATA_DIR_TRAINING = "C:/Users/user/Git Hub Projects/Machine_Learning_Projects/Image Classification/train"
DATA_DIR_VALIDATION = "C:/Users/user/Git Hub Projects/Machine_Learning_Projects/Image Classification/test"
img_size =(224,224)
batch_size = 32
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR_TRAINING,
    seed=42,
    image_size=img_size,
    batch_size = batch_size
)
train_ds
```

```python
val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR_VALIDATION,
    seed=42,
    image_size=img_size,
    batch_size = batch_size
)
val_ds
```

```python
class_names = train_ds.class_names
num_classes = len (class_names)
num_classes
```

```python
model = models.Sequential([
    layers.Rescaling(1./255,input_shape = img_size + (3,)),
    layers.Conv2D(16,3,activation="relu"),
    layers.MaxPool2D(),
    layers.Conv2D(32,3,activation="relu"),
    layers.MaxPool2D(),
    layers.Conv2D(64,3,activation="relu"),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(64,activation="relu"),
    layers.Dropout(0.5),
    layers.Dense(num_classes,activation="softmax")
])
```

```python
model.compile (
    optimizer ="adam",
    loss = "sparse_categorical_crossentropy",
    metrics = ["accuracy"]
)
```

```python
history = model.fit(
    train_ds,
    validation_data = val_ds,
    epochs = 5
)
```
